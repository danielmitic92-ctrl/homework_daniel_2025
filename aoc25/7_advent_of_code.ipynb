```python
import numpy as np
```

```python
manifold=np.array([list(x) for x in input_data])
```

```python
#determine initial beam
initial_beam_postion=np.where(manifold=="S")[1]
manifold[1][initial_beam_postion]="|"
```

```python
##function for index positions of incoming beams
current_line=2
incoming_beams=np.where(manifold[current_line-1]=="|")
```

```python
if manifold[current_line][(incoming_beams[0])]=="^":
    manifold[current_line,(incoming_beams[0])+1]="|"
    manifold[current_line,(incoming_beams[0])-1]="|"

if manifold[current_line][(incoming_beams[0])]==".":
    manifold[current_line,(incoming_beams[0])]="|"
```

```python
for current_line in range(2,len(manifold)):
    incoming_beams=np.where(manifold[current_line-1]=="|")
    
    if manifold[current_line][(incoming_beams[0] [0])]=="^":
        manifold[current_line,(incoming_beams[0] [0])+1]="|"
        manifold[current_line,(incoming_beams[0] [0])-1]="|"

    if manifold[current_line][(incoming_beams[0] [0])]==".":
        manifold[current_line,(incoming_beams[0] [0])]="|"
        
    
```
