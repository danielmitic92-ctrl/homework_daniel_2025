```python
#1 import input file
import util
file_path='/Users/danielmitic/Downloads/rosalind_splc (1).txt'
fasta=util.read_file_into_lines(file_path)
```

```python
fasta
```

```python
##preprocesing of input file
import re
sequence="".join(fasta) #join list to one string
sequence=sequence.split(">")[1:]#split along ID, exluding first line 

only_sequences=[]
for current_sequence in sequence:
    single_sequence=re.split('(\d+)', current_sequence)
    single_sequence=single_sequence[2]
    only_sequences.append(single_sequence)
```

```python
original_sequence=only_sequences[0]
introns=only_sequences[1:]
```

```python
#remove introns
for intron in introns:
    original_sequence=original_sequence.replace(intron,"")
```

```python
original_sequence
```

```python
#trancribe to rna
rna=original_sequence.replace("T","U")
```

```python
#as a function
import util
codon_aa_table=util.load_codon_table()
protein=util.translate_protein(rna,codon_aa_table)
```
