```python
#1 import input file
import util
file_path='/Users/danielmitic/Downloads/rosalind_lcsm (3).txt'
fasta=util.read_file_into_lines(file_path)
```

```python
#find the shortest seqeuence

##preprocesing of input file
import re
sequence="".join(fasta) #join list to one string
sequence=sequence.split(">")[1:]#split along ID, exluding first line 
only_sequences=[]

for current_sequence in sequence:
    single_sequence=re.split('(\d+)', current_sequence)
    single_sequence=single_sequence[2]
    only_sequences.append(single_sequence)
shortest_sequence=min(only_sequences)
```

```python
# Generate all substrings
substrings = [shortest_sequence[i:j] for i in range(len(shortest_sequence)) for j in range(i+1, len(shortest_sequence)+1)]
all_motifs=sorted(substrings, key=len)[::-1] #get all motifs
```

```python
##find all substrings in shortest string from largest to smallest when searching from back

for motif in all_motifs:
## search for motif in list of other strings
    counter=0
    for sequence in only_sequences:
        if motif not in sequence:       
            counter=0
            break
        else:
            counter+=1
            if counter==len(only_sequences):
                print(motif)
                break
```
