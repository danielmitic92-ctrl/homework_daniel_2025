```python
seq="ABCDEFG"
```

```python
subset=3
```

```python
seq=seq*subset
```

```python
from itertools import permutations

perm = permutations(seq,subset)
all_combinations=[]
for i in perm:
    all_combinations.append(i)

#remove duplicates
#set(all_combinations)
```

```python
all_combinations=list(set(all_combinations))
```

```python
all_combinations.sort()
```

```python
for combo in all_combinations:
    for i in combo:
        print(i,end="")
    print()
    
```
