```python
#preprocesing of input file
import re
sequence="".join(fasta) #join list to one string
sequence=sequence.split(">")[1:]#split along ID, exluding first line 
```

```python
##create empty matrix
import numpy as np
single_sequence=re.split('(\d+)', sequence[1])
single_sequence=single_sequence[2]
shape = (4, len(single_sequence)) 
arr = np.zeros(shape,dtype=int)
```

```python
arr
```

```python
#create profile matrix
for current_sequence in sequence:
    single_sequence=re.split('(\d+)', current_sequence)
    single_sequence=single_sequence[2]
    for position in range(len(single_sequence)):
        if single_sequence[position]=="A":
            arr[0,position]+=1

        if single_sequence[position]=="C":
            arr[1,position]+=1

        if single_sequence[position]=="G":
            arr[2,position]+=1

        if single_sequence[position]=="T":
            arr[3,position]+=1
```

```python
arr
```

```python
consensus=""
for current_position in range(len(arr[0,:])):
    highest_count=np.max(arr[:,current_position])
    ##look the highest sum is located
    index_highest_count=np.where(arr[:,current_position] == highest_count)
    if index_highest_count[0][0]==0:
        consensus=consensus+"A"
    if index_highest_count[0][0]==1:
        consensus=consensus+"C"
    if index_highest_count[0][0]==2:
        consensus=consensus+"G"
    if index_highest_count[0][0]==3:
        consensus=consensus+"T"
```
