```python
number=6
import numpy as np

seq = np.arange(1, number+1).tolist()
print(seq)
```

```python
from itertools import permutations

perm = permutations(seq)
all_combinations=[]
for i in perm:
    #print(i)
    all_combinations.append(i)
```

```python
print(len(all_combinations))
for combination in all_combinations:
    for i in combination:
        print(i, end=" ")
    print()
```
